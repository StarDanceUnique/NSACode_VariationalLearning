```python
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
f_nnsa = lambda gamma: 1 - 5 * gamma**2 
f_SC = lambda gamma: 1 - 3 * gamma**2 
f_CC = lambda gamma: 1 - 1.75*gamma**2

gamma = np.linspace(0, 0.2, 30)

fs = 20
plt.figure(figsize=(6, 4))
plt.plot(gamma, f_SC(gamma), label='SC', lw=3)
plt.plot(gamma, f_CC(gamma), label='PC', lw=3)
plt.plot(gamma, f_nnsa(gamma), label='LNCY', lw=3)
plt.xlabel(r'$\gamma$', fontsize=fs+2)
plt.ylabel(r'$F$', fontsize=fs+2)
plt.legend(fontsize=fs-7)
# Set x-axis ticks to 0, 0.05, 0.1, 0.15, 0.2
plt.xticks(np.linspace(0, 0.2, 5))
# Set y-axis ticks based on the function's value range, assumed to be around 0.8 to 1.0
plt.yticks(np.linspace(0.8, 1.0, 5))

plt.xticks(fontsize=fs-4)
plt.yticks(fontsize=fs-4)
plt.grid(True, which='both', linestyle='--', alpha=0.3)
ax = plt.gca()
xtick_fontsize = ax.xaxis.get_ticklabels()[0].get_size()
ytick_fontsize = ax.yaxis.get_ticklabels()[0].get_size()

print(f"X-axis tick label font size: {xtick_fontsize}")
print(f"Y-axis tick label font size: {ytick_fontsize}")
#plt.tight_layout()

filename = f'figure1b.pdf'
folder = 'figures'
filepath = os.path.join(folder, filename)
plt.savefig(filepath, bbox_inches='tight')
```
